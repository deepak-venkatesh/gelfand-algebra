```python
#Problem 257a

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-8, 5, 1000)

# Standard parabola with vertex at (0, 0)
y_standard = 2*x**2

# Same parabola shifted 3 units left and 5 units up
y_shifted = 2*(x + 3)**2 + 5

plt.figure(figsize=(10, 7))

plt.plot(
    x,
    y_standard,
    color="blue",
    linewidth=2,
    label=r"$y=2x^2$"
)

plt.plot(
    x,
    y_shifted,
    color="red",
    linewidth=2,
    label=r"$y=2(x+3)^2+5$"
)

# Original coordinate axes
plt.axhline(y=0, color="black", linewidth=1)
plt.axvline(x=0, color="black", linewidth=1)

# New coordinate axes through (-3, 5)
plt.axvline(
    x=-3,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $Y$-axis: $x=-3$"
)

plt.axhline(
    y=5,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $X$-axis: $y=5$"
)

# Mark the two vertices
plt.scatter([0], [0], color="blue", zorder=5)
plt.scatter([-3], [5], color="red", zorder=5)

plt.annotate(
    r"old vertex $(0,0)$",
    xy=(0, 0),
    xytext=(1, 5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"new vertex $(-3,5)$",
    xy=(-3, 5),
    xytext=(-7, 10),
    arrowprops=dict(arrowstyle="->")
)

plt.xlim(-8, 5)
plt.ylim(-5, 35)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Translation of an upward-opening parabola")

plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
# Problem 267 d

import numpy as np
import matplotlib.pyplot as plt

# x-values
x = np.linspace(-2, 2, 1000)

# Function
y = x**4 - x**2

# Important points
roots_x = [-1, 0, 1]
roots_y = [0, 0, 0]

min_x = [-1 / np.sqrt(2), 1 / np.sqrt(2)]
min_y = [-0.25, -0.25]

# Create figure
plt.figure(figsize=(10, 7))

# Plot curve
plt.plot(
    x,
    y,
    color="blue",
    linewidth=2,
    label=r"$y=x^4-x^2$"
)

# Coordinate axes
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

# Mark roots
plt.scatter(
    roots_x,
    roots_y,
    color="red",
    s=60,
    zorder=5
)

# Mark minima
plt.scatter(
    min_x,
    min_y,
    color="green",
    s=60,
    zorder=5
)

# Root labels
plt.annotate(
    r"$(-1,0)$",
    xy=(-1, 0),
    xytext=(-1.7, 0.35),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$(0,0)$",
    xy=(0, 0),
    xytext=(0.2, 0.45),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$(1,0)$",
    xy=(1, 0),
    xytext=(1.15, 0.35),
    arrowprops=dict(arrowstyle="->")
)

# Minimum labels
plt.annotate(
    r"$\left(-\frac{1}{\sqrt{2}},-\frac{1}{4}\right)$",
    xy=(-1 / np.sqrt(2), -0.25),
    xytext=(-1.9, -0.75),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$\left(\frac{1}{\sqrt{2}},-\frac{1}{4}\right)$",
    xy=(1 / np.sqrt(2), -0.25),
    xytext=(0.8, -0.75),
    arrowprops=dict(arrowstyle="->")
)

# Local maximum label
plt.annotate(
    r"local maximum $(0,0)$",
    xy=(0, 0),
    xytext=(-0.9, 0.85),
    arrowprops=dict(arrowstyle="->")
)

# Limits
plt.xlim(-2, 2)
plt.ylim(-1, 4)

# Labels
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Graph of $y=x^4-x^2$")

# Grid and legend
plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
# Problem 267 e

import numpy as np
import matplotlib.pyplot as plt

# x-values
x = np.linspace(0, 5, 1000)

# Polynomial
y = (x - 1)*(x - 2)*(x - 3)*(x - 4)

# Roots
roots_x = [1, 2, 3, 4]
roots_y = [0, 0, 0, 0]

# Create the graph
plt.figure(figsize=(10, 7))

# Plot the curve
plt.plot(
    x,
    y,
    color="blue",
    linewidth=2,
    label=r"$y=(x-1)(x-2)(x-3)(x-4)$"
)

# Coordinate axes
plt.axhline(
    y=0,
    color="black",
    linewidth=1
)

plt.axvline(
    x=0,
    color="black",
    linewidth=1
)

# Mark the four roots
plt.scatter(
    roots_x,
    roots_y,
    color="red",
    s=60,
    zorder=5
)

# Annotate the roots
plt.annotate(
    r"$(1,0)$",
    xy=(1, 0),
    xytext=(0.35, 2.5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$(2,0)$",
    xy=(2, 0),
    xytext=(1.55, -3.5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$(3,0)$",
    xy=(3, 0),
    xytext=(3.1, -3.5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"$(4,0)$",
    xy=(4, 0),
    xytext=(4., 4.2),
    arrowprops=dict(arrowstyle="->")
)

# Viewing window
plt.xlim(0, 5)
plt.ylim(-5, 25)

# Labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"Graph of $y=(x-1)(x-2)(x-3)(x-4)$")

# Grid and legend
plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```
