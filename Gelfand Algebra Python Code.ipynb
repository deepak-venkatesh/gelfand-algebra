```python
# General plottoing of function

import numpy as np
import matplotlib.pyplot as plt

x1 = np.linspace(-5, -0.05, 500)
x2 = np.linspace(0.05, 5, 500)

plt.figure(figsize=(8,5))

plt.plot(x1, (x1 - 1)/x1**2)
plt.plot(x2, (x2 - 1)/x2**2)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid(True)

plt.xlabel("x")
plt.ylabel("a")
plt.title(r"$a=\frac{x-1}{x^2}$")

plt.show()
```

```python
#Problem 255

import numpy as np
import matplotlib.pyplot as plt

# The function is undefined at x = 0,
# so make two separate sets of x-values.
x_left = np.linspace(-8, -0.03, 2000)
x_right = np.linspace(0.03, 10, 3000)

# Function a = (x - 1) / x^2
a_left = (x_left - 1) / x_left**2
a_right = (x_right - 1) / x_right**2

# Create the graph
plt.figure(figsize=(10, 7))

# Plot both branches
plt.plot(
    x_left,
    a_left,
    linewidth=2,
    label=r"$a=\frac{x-1}{x^2}$"
)

plt.plot(
    x_right,
    a_right,
    linewidth=2
)

# Coordinate axes
plt.axhline(
    y=0,
    color="black",
    linewidth=1.2,
    label=r"$a=0$"
)

plt.axvline(
    x=0,
    color="black",
    linewidth=1.2
)

# Several horizontal dashed lines
plt.axhline(
    y=0.25,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=0.25$"
)

plt.axhline(
    y=0.10,
    color="orange",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=0.10$"
)

plt.axhline(
    y=-1,
    color="green",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-1$"
)

plt.axhline(
    y=-4,
    color="purple",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-4$"
)

# Mark the maximum point
plt.scatter(
    [2],
    [0.25],
    color="red",
    zorder=5
)

plt.annotate(
    r"maximum $(2,\;0.25)$",
    xy=(2, 0.25),
    xytext=(3.0, 0.8),
    arrowprops=dict(arrowstyle="->")
)

# Mark x = 1, where a = 0
plt.scatter(
    [1],
    [0],
    color="black",
    zorder=5
)

plt.annotate(
    r"$(1,0)$",
    xy=(1, 0),
    xytext=(1.4, -0.7),
    arrowprops=dict(arrowstyle="->")
)

# Viewing window
plt.xlim(-8, 10)
plt.ylim(-6, 2)

# Axis labels and title
plt.xlabel("x")
plt.ylabel("a")
plt.title(r"Horizontal lines intersecting $a=\frac{x-1}{x^2}$")

# Grid, legend, and display
plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
#Problem 256a

import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-3, 4, 1000)

# Function a = x^2 - x
a = x**2 - x

# Create figure
plt.figure(figsize=(10, 7))

# Plot the parabola
plt.plot(
    x,
    a,
    linewidth=2,
    label=r"$a=x^2-x$"
)

# Coordinate axes
plt.axhline(
    y=0,
    color="black",
    linewidth=1.2,
    label=r"$a=0$"
)

plt.axvline(
    x=0,
    color="black",
    linewidth=1.2
)

# Horizontal dashed lines
plt.axhline(
    y=2,
    color="green",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=2$"
)

plt.axhline(
    y=-0.10,
    color="orange",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-0.10$"
)

plt.axhline(
    y=-0.25,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-0.25$"
)

plt.axhline(
    y=-1,
    color="purple",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-1$"
)

# Vertex
plt.scatter(
    [0.5],
    [-0.25],
    color="red",
    zorder=5
)

plt.annotate(
    r"vertex $\left(\frac{1}{2},-\frac{1}{4}\right)$",
    xy=(0.5, -0.25),
    xytext=(1.1, -1.5),
    arrowprops=dict(arrowstyle="->")
)

# Viewing window
plt.xlim(-3, 4)
plt.ylim(-2, 6)

# Labels and title
plt.xlabel("x")
plt.ylabel("a")
plt.title(r"Horizontal lines intersecting $a=x^2-x$")

# Grid and legend
plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
#Problem 256b

import numpy as np
import matplotlib.pyplot as plt

# The function is undefined at x = 0,
# so use separate x-values on either side of zero
x_left = np.linspace(-8, -0.05, 2000)
x_right = np.linspace(0.05, 8, 2000)

# Function a = (x^2 + 1) / x
a_left = (x_left**2 + 1) / x_left
a_right = (x_right**2 + 1) / x_right

# Create the graph
plt.figure(figsize=(10, 7))

# Plot both branches
plt.plot(
    x_left,
    a_left,
    linewidth=2,
    label=r"$a=\frac{x^2+1}{x}=x+\frac{1}{x}$"
)

plt.plot(
    x_right,
    a_right,
    linewidth=2
)

# Coordinate axes
plt.axhline(
    y=0,
    color="black",
    linewidth=1.2
)

plt.axvline(
    x=0,
    color="black",
    linewidth=1.2
)

# Horizontal dashed lines

# Two positive roots
plt.axhline(
    y=3,
    color="green",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=3$"
)

# One repeated positive root
plt.axhline(
    y=2,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=2$"
)

# No real roots
plt.axhline(
    y=1,
    color="orange",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=1$"
)

# One repeated negative root
plt.axhline(
    y=-2,
    color="purple",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-2$"
)

# Two negative roots
plt.axhline(
    y=-3,
    color="blue",
    linestyle="--",
    linewidth=1.5,
    label=r"$a=-3$"
)

# Mark the turning points
plt.scatter(
    [1, -1],
    [2, -2],
    color="red",
    zorder=5
)

plt.annotate(
    r"minimum $(1,2)$",
    xy=(1, 2),
    xytext=(2, 3.8),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"maximum $(-1,-2)$",
    xy=(-1, -2),
    xytext=(-4, -3.8),
    arrowprops=dict(arrowstyle="->")
)

# Viewing window
plt.xlim(-8, 8)
plt.ylim(-8, 8)

# Labels and title
plt.xlabel("x")
plt.ylabel("a")
plt.title(r"Horizontal lines intersecting $a=x+\frac{1}{x}$")

# Grid and legend
plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
#Problem 257a

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-8, 5, 1000)

# Standard parabola with vertex at (0, 0)
y_standard = 2*x**2

# Same parabola shifted 3 units left and 5 units up
y_shifted = 2*(x + 3)**2 + 5

plt.figure(figsize=(10, 7))

plt.plot(
    x,
    y_standard,
    color="blue",
    linewidth=2,
    label=r"$y=2x^2$"
)

plt.plot(
    x,
    y_shifted,
    color="red",
    linewidth=2,
    label=r"$y=2(x+3)^2+5$"
)

# Original coordinate axes
plt.axhline(y=0, color="black", linewidth=1)
plt.axvline(x=0, color="black", linewidth=1)

# New coordinate axes through (-3, 5)
plt.axvline(
    x=-3,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $Y$-axis: $x=-3$"
)

plt.axhline(
    y=5,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $X$-axis: $y=5$"
)

# Mark the two vertices
plt.scatter([0], [0], color="blue", zorder=5)
plt.scatter([-3], [5], color="red", zorder=5)

plt.annotate(
    r"old vertex $(0,0)$",
    xy=(0, 0),
    xytext=(1, 5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"new vertex $(-3,5)$",
    xy=(-3, 5),
    xytext=(-7, 10),
    arrowprops=dict(arrowstyle="->")
)

plt.xlim(-8, 5)
plt.ylim(-5, 35)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Translation of an upward-opening parabola")

plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
#Problem 257b

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-8, 5, 1000)

# Standard downward-opening parabola
y_standard = -2*x**2

# Shifted 3 units left and 5 units up
y_shifted = -2*(x + 3)**2 + 5

plt.figure(figsize=(10, 7))

plt.plot(
    x,
    y_standard,
    color="blue",
    linewidth=2,
    label=r"$y=-2x^2$"
)

plt.plot(
    x,
    y_shifted,
    color="red",
    linewidth=2,
    label=r"$y=-2(x+3)^2+5$"
)

# Original coordinate axes
plt.axhline(y=0, color="black", linewidth=1)
plt.axvline(x=0, color="black", linewidth=1)

# New coordinate axes through (-3, 5)
plt.axvline(
    x=-3,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $Y$-axis: $x=-3$"
)

plt.axhline(
    y=5,
    color="red",
    linestyle="--",
    linewidth=1,
    label=r"new $X$-axis: $y=5$"
)

# Mark vertices
plt.scatter([0], [0], color="blue", zorder=5)
plt.scatter([-3], [5], color="red", zorder=5)

plt.annotate(
    r"old vertex $(0,0)$",
    xy=(0, 0),
    xytext=(1, -5),
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    r"new vertex $(-3,5)$",
    xy=(-3, 5),
    xytext=(-7, 10),
    arrowprops=dict(arrowstyle="->")
)

plt.xlim(-8, 5)
plt.ylim(-35, 15)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Translation of a downward-opening parabola")

plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```

```python
#Problem 258

import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-8, 2, 1000)

# Parabolas
y1 = 2*(x + 3)**2 + 5
y2 = 2*(x + 3)**2 + 10

plt.figure(figsize=(10, 7))

# Plot the curves
plt.plot(
    x, y1,
    color="blue",
    linewidth=2,
    label=r"$y=2(x+3)^2+5$"
)

plt.plot(
    x, y2,
    color="red",
    linewidth=2,
    label=r"$y=2(x+3)^2+10$"
)

# Coordinate axes
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

# Mark the vertices
plt.scatter([-3], [5], color="blue", s=50)
plt.scatter([-3], [10], color="red", s=50)

plt.annotate(
    r"$(-3,5)$",
    xy=(-3, 5),
    xytext=(-2.2, 3.5),
    color="blue",
    arrowprops=dict(arrowstyle="->", color="blue")
)

plt.annotate(
    r"$(-3,10)$",
    xy=(-3, 10),
    xytext=(-2.2, 11.5),
    color="red",
    arrowprops=dict(arrowstyle="->", color="red")
)

plt.xlim(-8, 2)
plt.ylim(-2, 30)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Vertical Translation of a Parabola")

plt.grid(True, alpha=0.35)
plt.legend()

plt.show()
```
